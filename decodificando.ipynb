# Manejo de excepciones en Python (try-except)


# Sentencia try, acompañada de except o finally, con orde: [], argumento opcional. * except puede aparecer múltiples veces.

```python
try:
    num1 = int(input("Introduce el primer número: "))
    num2 = int(input("Introduce el segundo número: "))

    resultado = num1 / num2

except ValueError as error:
    print("Error: debes introducir números enteros.")
    print("Detalle:", error)

except ZeroDivisionError as error:
    print("Error: no se puede dividir entre cero.")
    print("Detalle:", error)

else:
    print("La operación se ha realizado correctamente.")
    print("El resultado es:", resultado)

finally:
    print("Fin del programa.")
```
