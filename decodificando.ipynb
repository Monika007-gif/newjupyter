# Manejo de excepciones en Python (try-except)


# Sentencia try, acompañada de except o finally, con orde: [], argumento opcional. * except puede aparecer múltiples veces.

```python
try:
    num1 = int(input("Introduce el primer número: "))
    num2 = int(input("Introduce el segundo número: "))

    resultado = num1 / num2

except ValueError as error:
    print("Error: debes introducir números enteros.")
    print("Detalle:", error)

except ZeroDivisionError as error:
    print("Error: no se puede dividir entre cero.")
    print("Detalle:", error)

else:
    print("La operación se ha realizado correctamente.")
    print("El resultado es:", resultado)

finally:
    print("Fin del programa.")
```

```python
sep = '-' * 45 + '\n'
print(sep + 'EXCEPTION RAISED AND CAUGHT')
try:
    x = 'spam'[99]
except IndexError:
    print('except run') 
finally:
    print('finally run') 
print('after run')
```

```python
print(sep + 'NO EXCEPTION RAISED')
try:
    x = 'spam'[3]
except IndexError:
    print('except run') 
finally:
    print('finally run')
print('after run')
```

```python
print(sep + 'NO EXCEPTION RAISED, WITH ELSE')
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
else:
    print('else run')
finally:
    print('finally run')
print('after run')
```

```python
print(sep + 'EXCEPTION RAISED BUT NOT CAUGHT')
try:
    x = 1 / 0
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

```python
try:
    x = 1 / 0

except IndexError:
    print("except run")

finally:
    print("finally run")

print("after run")
```

# try intenta ejecutar 1/0
# Python produce un ZeroDivisionError, porque no se puede dividir entre cero.
# except IndexError no captura el error, porque espera IndexError
# finally sí se ejecuta e imprime finally run
# after run no se ejecuta porque la excepción sigue sin estar controlada.
