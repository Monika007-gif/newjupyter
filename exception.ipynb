# Try contiene el código que puede producir errores, except captura los indicados, else se ejecuta solo cuando no hay errores y finally se ejecuta siempre, aunque el error no haya sido capturado.

```python
try:
    x = 'spam'[99]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

# try intenta acceder a la posición 99 de la palabra spam, como no existe, hay error IndexError.
# except captura el error y muestra except run, finally se ejecuta siempre y muestra finally run, continúa el programa y muestra after run.

```python
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```
